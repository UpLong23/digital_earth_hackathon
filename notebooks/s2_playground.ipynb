Import openeo and authenticate

```python
import openeo

backend_url = "https://openeo.dataspace.copernicus.eu"
connection = openeo.connect(backend_url)

connection.authenticate_oidc()
```

```python
collections = connection.list_collection_ids()
sentinel_collections = [c for c in collections if "SENTINEL1" in c or "SENTINEL2" in c]
print(sentinel_collections)
```

# Important features

- Vegetation density (S2)
- Algea bloom in run off water streams (S2)
- Loding (S1?)
- Biomass/plantheight (S1?)

```python
# Set your bounding box (west, south, east, north) in WGS84 (EPSG:4326)
# Sweden (according to gemini 3.1 pro)
spatial_extent = {
    "west": 10.95,
    "south": 55.33,
    "east": 24.17,
    "north": 69.07
}

temporal_extent = ["2023-06-01", "2023-06-30"]

bands = ["B02", "B03", "B04", "B08"]

# SENTINEL2_L1C, top-of-atmosphere
# SENTINEL2_L2A, bottom of atmosphere

s2_cube = connection.load_collection(
    "SENTINEL2_L2A",
    spatial_extent=spatial_extent,
    temporal_extent=temporal_extent,
    bands=bands
)

print("Successfully created datacube:")
print(s2_cube)
```
